```python
import pandas as pd

df = pd.read_csv("_datasets_new_raw\\buchwaldhartwig3.csv")

df
```

```python
df_smiles = df.filter(like="smiles")
df_smiles
```

```python
df_smiles_reordered = (
    df_smiles.rename(columns = {
        "product_smiles": "product",
        "input_metal_and_ligand_smiles": "catalyst",
        "input_amine_smiles": "reactant_1",
        "input_aryl_halide_smiles": "reactant_2",
        "input_base_smiles": "base",
        "input_solvent_smiles": "solvent",
    })
    .loc[:, [
        "reactant_1",
        "reactant_2",
        "catalyst",
        "product",
        "base",
        "solvent",
    ]]
)

df_smiles_reordered
```

```python
def _before_pipe(val):
    if pd.isna(val):
        return val
    if isinstance(val, str):
        return val.split("|", 1)[0].strip()
    return val

for c in df_smiles_reordered.columns:
    if df_smiles_reordered[c].dtype == object:
        df_smiles_reordered[c] = df_smiles_reordered[c].apply(_before_pipe)

df_smiles_reordered
```

```python
df_smiles_reordered.loc[df_smiles_reordered["additive"] == "CS(=O)C", "additive"] = pd.NA

df_smiles_reordered
```

```python
df_smiles_reordered = df_smiles_reordered.copy()

df_smiles_reordered["reactants"] = df_smiles_reordered[["reactant_1", "reactant_2"]].apply(
    lambda row: ".".join([x for x in row if pd.notna(x)]),
    axis=1
)

df_smiles_reordered["reagents"] = df_smiles_reordered[["base", "catalyst", "solvent"]].apply(
    lambda row: ".".join([x for x in row if pd.notna(x)]),
    axis=1
)

cols = ["reactants", "reagents", "product"] + [
    c for c in df_smiles_reordered.columns if c not in ["reactants", "reagents", "product"]
]

df_smiles_reordered = df_smiles_reordered[cols]

df_smiles_reordered
```

```python
df_smiles_reordered.to_csv("_datasets_new_cleaned\\buchwald_hartwig_3_cleaned.csv", index=False)
```

```python
from rdkit import Chem

def canonicalize_smiles(smiles):
    """Convert a SMILES string to its canonical form."""
    if pd.isna(smiles):
        return smiles
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is not None:
            return Chem.MolToSmiles(mol)
        return smiles
    except:
        return smiles

```

```python
df1 = pd.read_csv("_datasets_new_cleaned\\minisci_cleaned.csv")
```

```python
# Apply canonicalization to all object columns in df1
for col in df1.columns:
    if df1[col].dtype == object:
        df1[col] = df1[col].apply(canonicalize_smiles)

df1
```

```python
df1.to_csv("_datasets_new_cleaned\\minisci_cleaned_canonical.csv", index=False)
```
