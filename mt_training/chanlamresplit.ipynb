```python
import pandas as pd
```

```python
df_all = []

df_train = pd.read_csv("../_datasets_unprocessed/chanlam_dataset_unprocessed/chanlam_train.csv")
df_val = pd.read_csv("../_datasets_unprocessed/chanlam_dataset_unprocessed/chanlam_val.csv")
df_test = pd.read_csv("../_datasets_unprocessed/chanlam_dataset_unprocessed/chanlam_test.csv")

df_all.append(df_train)
df_all.append(df_val)
df_all.append(df_test)

new_df = pd.concat(df_all, ignore_index=True)

new_df["input_base"] = new_df["input_base"].str.replace("NoBase", "")
```

```python
from rdkit import Chem

def canonicalise_smiles(smiles):
    """Convert SMILES to canonical form, return empty string if invalid."""
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is not None:
            return Chem.MolToSmiles(mol)
        return ""
    except:
        return ""

# Canonicalise all SMILES columns
new_df["input_boronic_acid"] = new_df["input_boronic_acid"].apply(canonicalise_smiles)
new_df["input_sulfonamide"] = new_df["input_sulfonamide"].apply(canonicalise_smiles)
new_df["input_catalyst"] = new_df["input_catalyst"].apply(canonicalise_smiles)
new_df["input_base"] = new_df["input_base"].apply(canonicalise_smiles)
new_df["input_solvent"] = new_df["input_solvent"].apply(canonicalise_smiles)
new_df["product_1_canonical_smiles"] = new_df["product_1_canonical_smiles"].apply(canonicalise_smiles)
new_df["product_2_canonical_smiles"] = new_df["product_2_canonical_smiles"].apply(canonicalise_smiles)


# Save to CSV
new_df.to_csv("../_datasets_unprocessed/chanlam_dataset_unprocessed/mergeddata.csv", index=False)
```

```python
print("Unique Boronic Acids:", new_df["input_boronic_acid"].nunique())
print("Unique Sulfonamides:", new_df["input_sulfonamide"].nunique())
print("Unique Bases:", new_df["input_base"].nunique())
print("Unique Catalysts:", new_df["input_catalyst"].nunique())
print("Unique Solvents:", new_df["input_solvent"].nunique())
```

```python
# List all unique elements and their counts for each column
for column in new_df.columns:
    print(f"\n{'='*60}")
    print(f"Column: {column}")
    print(f"{'='*60}")
    value_counts = new_df[column].value_counts()
    print(f"Total unique values: {len(value_counts)}\n")
    print(value_counts)
```

```python
# Sort the dataframe by input_sulfonamide column
new_df_sorted = new_df.sort_values(by='input_sulfonamide')

print(new_df_sorted)
```
