```python
import os
from ord_schema.message_helpers import load_message
from ord_schema.proto import dataset_pb2

file_path = "_datasets_new/data/pfizer39k.pb.gz"

if not os.path.exists(file_path):
    print("❌ Error: File path does not exist!")
else:
    try:
        dataset = load_message(file_path, dataset_pb2.Dataset)
        print(f"✅ Success! Found Dataset: {dataset.name}")
        print(f"📊 Number of reactions inside file: {len(dataset.reactions)}")
    except Exception as e:
        print(f"❌ Failed to parse file structure. Error: {e}")
```
